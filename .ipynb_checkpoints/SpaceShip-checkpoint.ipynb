```python
orig_df["Group"] = orig_df["PassengerId"].str.split("_").str[0].astype(int)
orig_df["GroupSize"] = orig_df.groupby("Group")["Group"].transform("count")

orig_df = orig_df.drop(['PassengerId','Name'], axis=1)
orig_df['Transported'] = orig_df['Transported'].astype(int)
orig_df = orig_df.drop(['Transported_int'], axis=1)
orig_df.head(5)
```

```python
train_df=orig_df.drop(['VIP', 'Destination'], axis=1)
train_df.head(5)
```

```python
train_df[["Deck", "Cabin_num", "Side"]] = train_df["Cabin"].str.split("/", expand=True)
train_df.head(5)
```

```python
try:
    train_df = train_df.drop('Cabin', axis=1)
except KeyError:
    print("Field does not exist")
```

```python
#The SimpleImputer class provides basic strategies for imputing missing values.
from sklearn.impute import SimpleImputer
```

```python
# Числовые — медианой
num_cols = train_df.select_dtypes(include=[np.number]).columns
train_df[num_cols] = train_df[num_cols].fillna(train_df[num_cols].median())

# Категориальные — модой
cat_cols = train_df.select_dtypes(include=["object"]).columns
for col in cat_cols:
    train_df[col] = train_df[col].fillna(train_df[col].mode()[0])
```

```python
train_df.isna().sum()  
```

```python
train_df = pd.get_dummies(train_df, columns=["HomePlanet", "Side"], drop_first=True)
train_df["CryoSleep"] = train_df["CryoSleep"].astype(int)
```

```python
train_df.head()
```

```python
X = train_df.drop(columns=["Group","Transported"])
y = train_df["Transported"]
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder()
train_df["Deck_encoded"] = encoder.fit_transform(train_df[["Deck"]])
```

```python
train_df=train_df.drop(columns=["Deck"])
train_df.head()

```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import accuracy_score

X = train_df.drop(columns=["Transported"])
y = train_df["Transported"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

clf = HistGradientBoostingClassifier(
    max_iter=500,
    learning_rate=0.05,
    max_depth=6,
    l2_regularization=0.1,
    random_state=42
)
clf.fit(X_train, y_train)

print("Accuracy:", accuracy_score(y_test, clf.predict(X_test)))

# Кросс-валидация
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(clf, X, y, cv=cv, scoring="accuracy", n_jobs=-1)
print(f"CV: {scores.mean():.4f} ± {scores.std():.4f}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

clf_with_scaler = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)
clf_with_scaler.fit(X_train, y_train)

y_pred_scaled = clf_with_scaler.predict(X_test)
acc_scaled = accuracy_score(y_test, y_pred_scaled)
print(f"Со scaler:    accuracy = {acc_scaled:.4f}")
```
